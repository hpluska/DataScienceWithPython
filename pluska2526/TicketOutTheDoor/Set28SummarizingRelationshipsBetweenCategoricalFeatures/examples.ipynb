```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency   

npi = pd.read_csv("data.csv", delimiter= "\t")
display(npi.head())

# Creates a cross table of counts
contingency_table = pd.crosstab(npi['influence'], npi['leader'])

chi2, p, dof, expected = chi2_contingency(contingency_table)
# Displays the expected counts
display(np.round(expected))
display(np.round(chi2))



# chi2, pval, dof, expected = chi2_contingency(influence_leader_freq)
# print(np.round(expected))
```
