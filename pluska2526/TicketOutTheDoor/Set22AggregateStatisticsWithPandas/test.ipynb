```python
import pandas as pd
import numpy as np
import random
states = [
    "AL", "AK", "AZ", "AR", "CA", "CO", "CT", "DE", "FL", "GA",
    "HI", "ID", "IL", "IN", "IA", "KS", "KY", "LA", "ME", "MD",
    "MA", "MI", "MN", "MS", "MO", "MT", "NE", "NV", "NH", "NJ",
    "NM", "NY", "NC", "ND", "OH", "OK", "OR", "PA", "RI", "SC",
    "SD", "TN", "TX", "UT", "VT", "VA", "WA", "WV", "WI", "WY"
]

customers = pd.read_csv("Orders.csv")

customers['age'] = np.random.randint(20, 51, size=len(customers))
random_states = []
for s in range(len(customers)):
    random_states.append(random.choice(states))

customers['state'] = random_states

# newdf = customers.drop(columns = ['email','shoe_type','shoe_material','shoe_color','price'])
# display(newdf.head())

# newdf.to_csv("customers.csv")

# print(customers.age.median())
# print(customers.state.nunique())
# print(customers.state.unique())
# display(customers.head(5))
# display(customers.groupby('shoe_type').price.mean())
# display(customers.groupby('shoe_type').price.mean().reset_index())
# counts = customers.state.value_counts()
# display(counts)

# display(customers.groupby('shoe_type').age.apply(lambda x: np.percentile(x, 75)))

shoe_counts = customers.groupby(["shoe_type", "shoe_color"]).price.apply(lambda x: np.percentile(x, 30)).reset_index()
# display(shoe_counts)

shoe_counts.pivot(columns='shoe_type',
         index='shoe_color',
         values= 'price')
# display(shoe_counts)

shoe_counts2 = customers.groupby(["shoe_type", "age"]).id.count().reset_index()
shoe_counts2.pivot(columns = 'shoe_type', 
    index = 'age',
    values = 'id')

results = customers[(customers['shoe_type'] == 'ballet flats')]
display(results)
```
