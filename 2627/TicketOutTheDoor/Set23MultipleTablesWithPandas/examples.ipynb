```python
import pandas as pd

sales = pd.read_csv('sales.csv', delimiter = '\t')
targets = pd.read_csv('targets.csv', delimiter = '\t')
men_women = pd.read_csv('men_women.csv')


display(sales)
display(targets)
display(men_women)

combined = sales.merge(targets).merge(men_women)
display(combined)


results = combined[(combined.revenue > combined.target) & (combined.women > combined.men)]
display(results)
```

```python
customers = pd.DataFrame({
    "customer_id": ["C01", "C02"],
    "customer_name": ["Alex Kim", "Jordan Lee"]
})

products = pd.DataFrame({
    "product_id": ["P10", "P12"],
    "product_name": ["Laptop", "Mouse"],
    "product_price": [999.99, 25.99]
})

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": ["C01", "C01", "C02", "C01"],
    "product_id": ["P10", "P12", "P10", "P10"],
    "quantity": [1, 2, 1, 1]
})

# display(customers)
# display(products)
# display(orders)

full_orders = orders.merge(customers).merge(products)
display(full_orders)

customer_totals = full_orders.groupby("customer_name").apply(lambda x: (x.product_price * x.quantity).sum()).reset_index()

display(customer_totals)
```
