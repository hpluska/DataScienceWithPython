```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm






study_hours = pd.read_csv("study_hours.csv")

plt.scatter(study_hours["hours"], study_hours["score"])
plt.xlabel("Hours Studied")
plt.ylabel("Score")
plt.title("Study Hours vs. Score")
plt.show()

model = sm.OLS.from_formula('score ~ hours', data = study_hours)
results = model.fit()
print(results.params)

fitted_values = results.predict(study_hours)
residuals = study_hours.score - fitted_values
print(residuals.head())

plt.hist(residuals, bins = 10)
plt.xlabel("Residual")
plt.ylabel("Frequency")
plt.title("Histogram of Residuals")
plt.show()

plt.scatter(fitted_values, residuals)
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.title("residuals vs fitted values")
plt.show()


```
