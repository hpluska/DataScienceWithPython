```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
housing = pd.read_csv('housing.csv', delimiter = "\t")
display(housing.head())

plt.scatter(x = housing.price, y = housing.sqfeet)
plt.xlabel('Rental Price (USD)')
plt.ylabel('Area (Square Feet)')
plt.show()

cov_price_sqfeet = housing['price'].cov(housing['sqfeet'])
print(cov_price_sqfeet)
cov_mat_price_sqfeet = housing[['price', 'sqfeet']].cov()
print(cov_mat_price_sqfeet)

from scipy.stats import pearsonr
# returns two values as a tuple, the first value is the correlation
result = pearsonr(housing.price, housing.sqfeet)


corr_price_sqfeet = result[0]
p = result[1]

print(corr_price_sqfeet)
print(p)
```
