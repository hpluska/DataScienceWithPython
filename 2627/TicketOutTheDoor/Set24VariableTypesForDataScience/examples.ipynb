```python
import pandas as pd

cereal = pd.read_csv('cereal.csv')



cereal['shelf'] = cereal.shelf.astype('category')

cereal["shelf"] = cereal["shelf"].cat.rename_categories({
    3: "top",
    1: "bottom",
    2: "mid"
})

display(cereal.shelf.unique)

shelf_order = ['bottom', 'mid', 'top']
cereal["shelf"] = cereal["shelf"].cat.set_categories(shelf_order, ordered=True)
display(cereal.shelf.unique)

# display(cereal[cereal["shelf"] > "mid"])

display(cereal.type.unique())

cereal = pd.get_dummies(data = cereal, columns = ['type'])

display(cereal.head())
```
