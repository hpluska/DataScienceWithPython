```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
students = pd.read_csv('data.csv', delimiter = "\t")
display(students.head())

# Gets the scores of from GP
gp_students = students[students.school == 'GP']['G3']
# Gets the scores of from MS
ms_students = students[students.school == 'MS']['G3']

# Calculates the mean for students at both schools
gp_mean = gp_students.mean()
ms_mean = ms_students.mean()

# Calculates the median for students at both schools
gp_median = gp_students.median()
ms_median = ms_students.median()

# print(gp_mean, ms_mean)
# print(gp_median, ms_median)

mean_diff = gp_mean - ms_mean
median_diff = gp_median - ms_median

print(mean_diff, median_diff)

# sns.boxplot(data = students, x = 'school', y = 'G3', hue = 'school', legend = True)
# plt.show()


# plt.hist(gp_students , color="blue", label="GP", density=True, alpha=0.5)
# plt.hist(ms_students , color="red", label="MS", density=True, alpha=0.5)
# plt.legend()
# plt.show()

sns.boxplot(data = students, 
            x = 'Mjob', 
            y = 'G3', 
            hue='Mjob',
            legend=False
            )
plt.show()
```
