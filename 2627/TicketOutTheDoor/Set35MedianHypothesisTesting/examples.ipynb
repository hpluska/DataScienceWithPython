```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the cleaned income dataset
df = pd.read_csv("income.csv")

# Use the income column
income = df["INCTOT"]

# Calculate the population median from this dataset
population_median = income.median()
print("Population median:", population_median)

# Function to generate many sample medians
def generate_sample_medians(data, sample_size, num_samples):
    medians = []
    for _ in range(num_samples):
        sample = np.random.choice(data, size=sample_size, replace=False)
        medians.append(np.median(sample))
    return medians

# Settings
sample_size = 50

# Generate sample medians
medians_10 = generate_sample_medians(income, sample_size, 10)
medians_100 = generate_sample_medians(income, sample_size, 100)
medians_500 = generate_sample_medians(income, sample_size, 500)

# Make the three histograms
plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.hist(medians_10, bins=10)
plt.axvline(population_median, color='red', linestyle='--')
plt.title("10 Sample Medians")
plt.xlabel("Sample Median Income")
plt.ylabel("Count")

plt.subplot(1, 3, 2)
plt.hist(medians_100, bins=15)
plt.axvline(population_median, color='red', linestyle='--')
plt.title("100 Sample Medians")
plt.xlabel("Sample Median Income")
plt.ylabel("Count")

plt.subplot(1, 3, 3)
plt.hist(medians_500, bins=20)
plt.axvline(population_median, color='red', linestyle='--')
plt.title("500 Sample Medians")
plt.xlabel("Sample Median Income")
plt.ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the cleaned housing dataset
df = pd.read_csv("housing.csv")

# Use the home value column
home_values = df["VALUEH"]

# Calculate the population median from this dataset
population_median = home_values.median()
print("Population median home value:", population_median)

# Function to generate many sample medians
def generate_sample_medians(data, sample_size, num_samples):
    medians = []
    for _ in range(num_samples):
        sample = np.random.choice(data, size=sample_size, replace=False)
        medians.append(np.median(sample))
    return medians

# Settings
sample_size = 50

# Generate sample medians
medians_10 = generate_sample_medians(home_values, sample_size, 10)
medians_100 = generate_sample_medians(home_values, sample_size, 100)
medians_500 = generate_sample_medians(home_values, sample_size, 500)

# Use a common x-axis range for all three histograms
xmin = min(min(medians_10), min(medians_100), min(medians_500))
xmax = max(max(medians_10), max(medians_100), max(medians_500))

# Make the three histograms
plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.hist(medians_10, bins=10, range=(xmin, xmax))
plt.axvline(population_median, color='red', linestyle='--')
plt.title("10 Sample Medians")
plt.xlabel("Sample Median Home Value")
plt.ylabel("Count")
plt.xlim(xmin, xmax)

plt.subplot(1, 3, 2)
plt.hist(medians_100, bins=15, range=(xmin, xmax))
plt.axvline(population_median, color='red', linestyle='--')
plt.title("100 Sample Medians")
plt.xlabel("Sample Median Home Value")
plt.ylabel("Count")
plt.xlim(xmin, xmax)

plt.subplot(1, 3, 3)
plt.hist(medians_500, bins=20, range=(xmin, xmax))
plt.axvline(population_median, color='red', linestyle='--')
plt.title("500 Sample Medians")
plt.xlabel("Sample Median Home Value")
plt.ylabel("Count")
plt.xlim(xmin, xmax)

plt.tight_layout()
plt.show()
```

```python


income_data = pd.read_csv("income.csv")

def bootstrap_median(original_sample, label, replications):
    just_one_column = original_sample[label]
    medians = []

    for i in range(replications):
        bootstrap_sample = np.random.choice(just_one_column, size=len(just_one_column), replace=True)
        resampled_median = np.median(bootstrap_sample)
        medians.append(resampled_median)

    return medians

bootstrap_median(income_data,'INCTOT',500)


boot_medians = bootstrap_median(income_data, "INCTOT", 500)
original_sample_median = np.median(income_data["INCTOT"])

plt.figure(figsize=(10, 5))
plt.hist(boot_medians, bins=30)
plt.axvline(original_sample_median, linestyle="--", label=f"Original Sample Median = {original_sample_median:,.0f}")
plt.xlabel("Bootstrapped Sample Median Income")
plt.ylabel("Count")
plt.title("Distribution of 500 Bootstrapped Sample Medians")
plt.legend()
plt.show()

observed_median = 47500
count = sum(1 for median in boot_medians if median >= observed_median)
p_value = count / len(boot_medians)
print(p_value)
```
